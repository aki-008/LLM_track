```python
def as_numpy(arr):
    if HAS_GPU and hasattr(arr, "get"):
        return arr.get()
    if isinstance(arr, np.ndarray):
        return arr
    return np.array(arr)

def as_device(arr):
    if HAS_GPU and isinstance(arr, np.ndarray):
        return xp.asarray(arr)
    return arr

def print_device():
    print(f'Using: {DEVICE}')
```

```python
arr = as_numpy(arr)
arr.device
```

```python
arr = as_device(arr)
arr.device
```
